Boxplot for Outliers

```python
plt.figure()
sns.boxplot(
    x=df['kills'],
    width=0.4,
    linewidth=2,
    fliersize=5
)

plt.title("Boxplot of Kills Distribution")
plt.xlabel("Kills per Match")
sns.set_style("darkgrid")
sns.set_context("talk")

plt.tight_layout()
plt.show()
```

Average kills ≈ 14.87
Most players score around 15 kills per match
75% of players have 18 or fewer kills
Performance distribution is moderately spread with few high performers

Histogram – Kills

```python
plt.figure()
plt.hist(df["kills"], bins=15, edgecolor="black", alpha=0.8)
plt.title("Distribution of Kills")
plt.xlabel("Kills per Match")
plt.ylabel("Number of Players")
plt.tight_layout()
plt.show()
```

Most players have 12–18 kills
Average ≈ 15 kills
Very few players below 5 or above 25

Histogram – Accuracy

```python
plt.figure()
plt.hist(df["accuracy_percent"], bins=15, edgecolor="black", alpha=0.8)
plt.title("Distribution of Accuracy Percentage")
plt.xlabel("Accuracy (%)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

Mean accuracy ≈ 45%
Most players fall in 40–50% competitive range
Very low accuracy (<25%) is rare
High accuracy players (>60%) are limited

Histogram – Reaction Time

```python
plt.figure()
plt.hist(df["reaction_time_ms"], bins=15, edgecolor="black", alpha=0.8)
plt.title("Distribution of Reaction Time")
plt.xlabel("Reaction Time (ms)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

Average ≈ 220 ms
Fastest ≈ 118 ms
Lower reaction time generally indicates better performance
Majority of players fall between 200–240 ms

Boxplot – Performance Score

```python
plt.figure()
plt.boxplot(df["performance_score"],
            patch_artist=True,
            boxprops=dict(linewidth=2),
            medianprops=dict(linewidth=2))
plt.title("Boxplot of Performance Score")
plt.ylabel("Performance Score")
plt.tight_layout()
plt.show()
```

Median ≈ 220 ms
Most values between 200–240 ms
Some slower outliers above 300 ms

Scatter – Kills vs Performance Score

```python
plt.figure()
plt.scatter(df["kills"], df["performance_score"], alpha=0.6)
plt.title("Kills vs Performance Score")
plt.xlabel("Kills")
plt.ylabel("Performance Score")
plt.tight_layout()
plt.show()
```

Positive relationship
As accuracy increases → performance score increases

Scatter – Accuracy vs Win Probability

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df["accuracy_percent"], 
    df["win_probability"], 
    alpha=0.6
)

plt.title("Accuracy vs Win Probability", fontsize=14)
plt.xlabel("Accuracy (%)", fontsize=12)
plt.ylabel("Win Probability", fontsize=12)

plt.grid(True)
plt.tight_layout()
plt.show()
```

Points generally move upward
As accuracy increases, win probability also increases
Most data points cluster in:
    Accuracy: 40%–55%
    Win Probability: 0.85–1.0

Correlation Heatmap

```python
plt.figure()
corr = df.select_dtypes(include=[np.number]).corr()

plt.imshow(corr, interpolation="nearest")
plt.title("Correlation Heatmap")
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.colorbar()
plt.tight_layout()
plt.show()
```

+1 → Strong positive relationship
0 → No relationship
-1 → Strong negative relationship

Strong positive correlation: Kills ↔ Performance
Moderate correlation: Accuracy ↔ Win Probability
Likely negative correlation: Reaction Time ↔ Performance
Win probability appears skewed high (mean ≈ 91%)

```python
# Fill missing values in numerical columns using mean (average value)

df["sleep_hours"] = df["sleep_hours"].fillna(df["sleep_hours"].mean())  # Replace missing sleep hours with average
df["sleep_efficiency"] = df["sleep_efficiency"].fillna(df["sleep_efficiency"].mean())  # Replace missing sleep efficiency with average
df["postprandial_glucose"] = df["postprandial_glucose"].fillna(df["postprandial_glucose"].mean())  # Replace missing post-meal glucose with average
df["fasting_glucose"] = df["fasting_glucose"].fillna(df["fasting_glucose"].mean())  # Replace missing fasting glucose with average
df["bp_systolic"] = df["bp_systolic"].fillna(df["bp_systolic"].mean())  # Replace missing systolic BP with average
df["bp_diastolic"] = df["bp_diastolic"].fillna(df["bp_diastolic"].mean())  # Replace missing diastolic BP with average
```

## Missing Value Handling Technique

- **`fillna()`** → Used to replace missing (**NaN**) values in the dataset  

- **`mean()`** → Calculates the **average value** of a column  

---

## Method Used: Mean Imputation

This technique is known as **Mean Imputation**, where missing values are replaced with the **mean (average) of the respective column**.

### Benefits:
- Simple and easy to implement  
- Preserves dataset size  
- Works well for numerical data  

### Limitation:
- Can reduce data variability  
- May introduce bias if data is not randomly missing

## Correlation Heatmap Insights

```python
corr = df.select_dtypes(include=np.number).corr()
threshold = 0.5

# Remove diagonal
np.fill_diagonal(corr.values, np.nan)

# Keep only strong relationships
strong_corr = corr.where(corr.abs() > threshold)

# Drop empty rows/columns
strong_corr = strong_corr.dropna(how="all", axis=0).dropna(how="all", axis=1)

# Mask upper triangle
mask = np.triu(np.ones_like(strong_corr, dtype=bool))

plt.figure(figsize=(13, 9))

sns.heatmap(
    strong_corr,
    mask=mask,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    annot_kws={"size": 9},
    square=False,
    cbar_kws={"shrink": 0.8}
)

plt.title("Filtered Strong Correlation Heatmap", fontsize=16, fontweight="bold")
plt.xticks(rotation=45, ha="right", fontsize=10)
plt.yticks(rotation=0, fontsize=10)
plt.tight_layout()
plt.show()
```



Based on the correlation heatmap, several important relationships between variables can be observed:

---

### Strong Positive Correlations

- **Steps, Heart Points, Move Minutes, and Calories Burned** show **very strong positive correlation**  
  → This indicates that higher physical activity directly increases calorie burn and movement metrics.

- **Weight (kg) and BMI (~0.81)**  
  → Strong relationship, as BMI is calculated using weight and height.

- **Sleep Hours and Sleep Efficiency (~0.69)**  
  → Better sleep duration leads to improved sleep quality.

- **Fatigue Score and Cardiometabolic Risk (~0.82)**  
  → Higher fatigue is strongly linked with increased health risk.

---

### Strong Negative Correlations

- **Sleep Efficiency vs Fatigue Score (~ -0.66)**  
  → Better sleep reduces fatigue levels.

- **HRV vs Resting Heart Rate (~ -0.63)**  
  → Higher heart rate variability is associated with better cardiovascular health.

- **Activity Metrics vs Fatigue**  
  → Increased activity tends to reduce fatigue levels.

---

### Moderate Correlations

- **Blood Pressure (Systolic & Diastolic)** are positively correlated  
  → Indicates consistent cardiovascular measurements.

- **Glucose Levels (Fasting & Postprandial)** show moderate positive correlation  
  → Important indicators for metabolic health.

- **Body Temperature & Fatigue (~0.50)**  
  → Higher fatigue may slightly relate to elevated body temperature.

---

### Weak or No Correlation

- **Water Intake** shows very low correlation with most variables  
- **User ID and Date-related features** do not impact health metrics  
  → These are not useful for prediction models

---

### Key Takeaways

- **Physical activity is a major factor** influencing calories, heart rate, and overall health.
